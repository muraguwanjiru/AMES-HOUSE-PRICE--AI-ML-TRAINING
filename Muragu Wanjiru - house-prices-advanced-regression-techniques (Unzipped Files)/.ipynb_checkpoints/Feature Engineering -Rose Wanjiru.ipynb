# AMES Housing Price Prediction: Advanced Feature Engineering & Regularization

### Introduction

Predicting real estate valuations is one of the definitive challenges in predictive modeling, driven by complex feature relationships, structural patterns, and volatile pricing distributions. This project leverages the **Ames Housing Dataset** to build a high-performance regression pipeline capable of forecasting property values in Ames, Iowa. The core challenge of this analysis does not lie in simply fitting an algorithm, but rather in navigating the structural traps and data-entry nuances that commonly compromise baseline machine learning frameworks.

Most predictive pipelines fail by treating data entry omissions blindly or misinterpreting encoding math. For instance, default data parsers read structural code markers (such as `"NA"` for houses lacking a physical basement or pool) as missing errors, leading to improper statistical imputation. Similarly, category labels masquerading as numbers (such as `MSSubClass` dwelling types) are often processed as numerical ranges, distorting the model's coefficients. Furthermore, real estate datasets are naturally right-skewed; a handful of multi-million-dollar luxury estates create a long distribution tail that can wildly destabilize standard linear algorithms.

To address these vulnerabilities, this project establishes a rigorous, end-to-end framework built on advanced feature-reasoning principles. I systematically isolate structural attributes from genuine missing values, enforce custom ordinal rankings, and deploy **fold-isolated Target Encoding** to map categorical weights strictly within a 5-fold Cross-Validation structure. Finally, a natural log-transformation stabilizes target variance, while a **Lasso Regression (L1 Regularization)** model trims non-essential feature coefficients to absolute zero. This strategic pipeline completely eliminates data leakage, balances high-dimensional complexity, and produces an optimized, highly generalizable model.

## 7. Fix the Data Type Trap (`MSSubClass`)

The `MSSubClass` column contains numeric codes (like `20`, `60`, and `190`) that represent categorical housing types (e.g., 1-Story vs 2-Story). If left as integers, the machine learning model will mistakenly assume a mathematical relationship between the classes (e.g., that class 60 is three times larger than class 20). 

To prevent this distortion, we change  this column to a string data type (`object`) before any categorical encoding occurs.


```python
if 'MSSubClass' in df_train.columns:
    df_train['MSSubClass'] = df_train['MSSubClass'].astype(str)
print(df_train['MSSubClass'].dtype)
```

## 8. Convert Text Rankings to Ordered Numbers (Ordinal Mapping)

Many categorical text features in this dataset carry a natural logical progression (e.g., `Ex` for Excellent is distinctly better than `Gd` for Good). Using standard One-Hot Encoding here would destroy this inherent ranking and create unnecessary columns.

We map these features manually into an explicit numerical scale (e.g., 1 to 5). Houses with no basement or missing features are mapped to a baseline rank of `0`.


```python
quality_rating_map = {'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5}
basement_rating_map = {'NA': 0, 'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5}
basement_exposure_map = {'NA': 0, 'No': 1, 'Mn': 2, 'Av': 3, 'Gd': 4}
property_shape_map = {'IR3': 1, 'IR2': 2, 'IR1': 3, 'Reg': 4}
land_slope_map = {'Sev': 1, 'Mod': 2, 'Gtl': 3}
utilities_utility_map = {'ELO': 1, 'NoSeWa': 2, 'NoSewr': 3, 'AllPub': 4}

for col in ['ExterQual', 'ExterCond']:
    if col in df_train.columns:
        df_train[col] = df_train[col].map(quality_rating_map)
        
for col in ['BsmtQual', 'BsmtCond']:
    if col in df_train.columns:
        df_train[col] = df_train[col].map(basement_rating_map)
        
if 'BsmtExposure' in df_train.columns:
    df_train['BsmtExposure'] = df_train['BsmtExposure'].map(basement_exposure_map)

if 'LotShape' in df_train.columns:
    df_train['LotShape'] = df_train['LotShape'].map(property_shape_map)
    
if 'LandSlope' in df_train.columns:
    df_train['LandSlope'] = df_train['LandSlope'].map(land_slope_map)

if 'Utilities' in df_train.columns:
    df_train['Utilities'] = df_train['Utilities'].map(utilities_utility_map)

df_train[['ExterQual', 'BsmtQual', 'LotShape', 'LandSlope']].head()

```

## 9: Handle True Missing Values (Imputation)

Now that structural `NA` categories are safely accounted for, any remaining blank entries represent genuine data omissions or entry errors (like an unmeasured street connection for `LotFrontage`). 

We handle these true omissions by filling numeric gaps with the column's statistical median and replacing empty categorical fields with baseline markers like `"None"` or `0.0`.


```python
if 'LotFrontage' in df_train.columns:
    df_train['LotFrontage'] = pd.to_numeric(df_train['LotFrontage'], errors='coerce')
    df_train['LotFrontage'] = df_train['LotFrontage'].fillna(df_train['LotFrontage'].median())
    
if 'MasVnrType' in df_train.columns:
    df_train['MasVnrType'] = df_train['MasVnrType'].replace('', 'None')
    
if 'MasVnrArea' in df_train.columns:
    df_train['MasVnrArea'] = pd.to_numeric(df_train['MasVnrArea'], errors='coerce').fillna(0.0)
```

## ONE-HOT ENCODING
### 10.Convert Nominal Categories Using One-Hot Encoding

For the remaining textual columns that do not have an inherent logical order or ranking (such as `MSZoning`, `Neighborhood`, `Street`, and our corrected `MSSubClass`), we apply One-Hot Encoding. 

This process transforms each unique textual category into its own binary indicator column (containing `1` or `0`). We use `drop_first=True` to eliminate multi-collinearity by dropping one baseline category per feature, ensuring our dataset remains mathematically stable for modeling.


```python
import pandas as pd

categorical_cols = df_train.select_dtypes(include=['object']).columns.tolist()

df_encoded = pd.get_dummies(df_train, columns=categorical_cols, drop_first=True)

boolean_cols = df_encoded.select_dtypes(include=['bool']).columns
df_encoded[boolean_cols] = df_encoded[boolean_cols].astype(int)
```

## 12. K-Fold Cross-Validation & Model Training

To evaluate model stability across all 245 features, I implement a 5-fold cross-validation routine using Lasso Regression. 

To completely eliminate data leakage, I execute all preprocessing steps locally inside the loop:
* I fit `SimpleImputer` using only the active fold's training median to handle missing entries.
* I fit `StandardScaler` on the training fold to normalize data types before transformation.
* I deploy `LassoCV` which is a form of regularization for linear regression.It makes features to shrink the weight of useless columns down to absolute zero. By zeroing out those unhelpful features, it prevents the model from overfitting and leaves  a clean, optimized list of only the variables that truly drive house prices.



```python
from sklearn.model_selection import KFold
from sklearn.linear_model import LassoCV
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
import numpy as np

X = df_encoded.drop(columns=['SalePrice'])
y = df_encoded['SalePrice']

kf = KFold(n_splits=5, shuffle=True, random_state=42)

fold_train_scores = []
fold_val_scores = []

fold_num = 1
for train_index, val_index in kf.split(X):
    X_train_fold, X_val_fold = X.iloc[train_index], X.iloc[val_index]
    y_train_fold, y_val_fold = y.iloc[train_index], y.iloc[val_index]
    
    # 1. Clean missing values locally inside the fold to prevent data leakage
    imputer = SimpleImputer(strategy='median')
    X_train_clean = imputer.fit_transform(X_train_fold)
    X_val_clean = imputer.transform(X_val_fold)
    
    # 2. Scale features locally
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train_clean)
    X_val_scaled = scaler.transform(X_val_clean)
    
    # 3. Train the Lasso model
    model = LassoCV(cv=5, random_state=42, max_iter=10000)
    model.fit(X_train_scaled, y_train_fold)
    
    # 4. Evaluate performance
    train_r2 = model.score(X_train_scaled, y_train_fold)
    val_r2 = model.score(X_val_scaled, y_val_fold)
    
    fold_train_scores.append(train_r2)
    fold_val_scores.append(val_r2)
    
    print(f"Fold {fold_num} -> Training R²: {train_r2:.4f} | Validation R²: {val_r2:.4f}")
    fold_num += 1


print(f"Mean Training R² Score:   {np.mean(fold_train_scores):.4f}")
print(f"Mean Validation R² Score: {np.mean(fold_val_scores):.4f}")

```

## TARGET ENCODING
## 13.Feature Type Identification
I isolate the raw qualitative columns from the numeric columns. This step ensures that only the unordered text features are isolated for category mean pricing substitution, keeping the original numerical columns completely intact.


```python
import pandas as pd
import numpy as np

text_columns = df_train.select_dtypes(include=['object']).columns.tolist()


X_raw = df_train.drop(columns=['SalePrice'])
y = df_train['SalePrice']

print(f"Isolated {len(text_columns)} categorical features to target encode.")
```

## 14.K-Fold Loop with Local Encoding & Imputation

I execute a 5-fold cross-validation loop to isolate partitions. Inside the loop, I compute group price averages strictly from the active training slice and map them onto the validation slice. This guarantees zero data leakage before scaling.


```python
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

kf = KFold(n_splits=5, shuffle=True, random_state=42)

fold_train_inputs = []
fold_val_inputs = []
fold_train_targets = []
fold_val_targets = []

for train_index, val_index in kf.split(X_raw):
    X_train_fold = X_raw.iloc[train_index].copy()
    X_val_fold = X_raw.iloc[val_index].copy()
    y_train_fold = y.iloc[train_index]
    y_val_fold = y.iloc[val_index]
    
    for col in text_columns:
        category_means = y_train_fold.groupby(X_train_fold[col]).mean()
        global_mean = y_train_fold.mean()
        
        X_train_fold[col] = X_train_fold[col].map(category_means).fillna(global_mean)
        X_val_fold[col] = X_val_fold[col].map(category_means).fillna(global_mean)
        
    imputer = SimpleImputer(strategy='median')
    X_train_clean = imputer.fit_transform(X_train_fold)
    X_val_clean = imputer.transform(X_val_fold)
    
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train_clean)
    X_val_scaled = scaler.transform(X_val_clean)
    
    fold_train_inputs.append(X_train_scaled)
    fold_val_inputs.append(X_val_scaled)
    fold_train_targets.append(y_train_fold)
    fold_val_targets.append(y_val_fold)

print(f"Successfully processed and stored data partitions across all {kf.n_splits} folds.")
```

## 15.Lasso Model Fitting & Performance Evaluation

I train the Lasso Regression model on each of the isolated, target-encoded data partitions. This stage calculates individual fold scores and computes the final cross-validation metric to verify overall model stability.


```python
from sklearn.linear_model import LassoCV

fold_train_scores = []
fold_val_scores = []

for i in range(5):
    model = LassoCV(cv=5, random_state=42, max_iter=10000)
    model.fit(fold_train_inputs[i], fold_train_targets[i])
    
    train_r2 = model.score(fold_train_inputs[i], fold_train_targets[i])
    val_r2 = model.score(fold_val_inputs[i], fold_val_targets[i])
    
    fold_train_scores.append(train_r2)
    fold_val_scores.append(val_r2)
    
    print(f"Fold {i+1} -> Target-Encoded Train R²: {train_r2:.4f} | Validation R²: {val_r2:.4f}")

print(f"Mean Training R² Score:   {np.mean(fold_train_scores):.4f}")
print(f"Mean Validation R² Score: {np.mean(fold_val_scores):.4f}")

```

These metrics confirm that target encoding captures neighborhood and zoning trends effectively without bloating the dataset.

##  17.Log Transformation

To stabilize variance and neutralize luxury outliers, I apply a natural log transformation (`np.log1p`) to the target variable before splitting. Inside the cross-validation folds, I compute category target encodings strictly on the log-transformed training prices to ensure no data leakage occurs before feature scaling.


```python
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer


y_log = np.log1p(y)

kf = KFold(n_splits=5, shuffle=True, random_state=42)

fold_train_inputs = []
fold_val_inputs = []
fold_train_targets = []
fold_val_targets = []

for train_index, val_index in kf.split(X_raw):
    X_train_fold = X_raw.iloc[train_index].copy()
    X_val_fold = X_raw.iloc[val_index].copy()
    y_train_fold = y_log.iloc[train_index]
    y_val_fold = y_log.iloc[val_index]
  
    for col in text_columns:
        category_means = y_train_fold.groupby(X_train_fold[col]).mean()
        global_mean = y_train_fold.mean()
        
        X_train_fold[col] = X_train_fold[col].map(category_means).fillna(global_mean)
        X_val_fold[col] = X_val_fold[col].map(category_means).fillna(global_mean)
        
    imputer = SimpleImputer(strategy='median')
    X_train_clean = imputer.fit_transform(X_train_fold)
    X_val_clean = imputer.transform(X_val_fold)
    
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train_clean)
    X_val_scaled = scaler.transform(X_val_clean)
    
    fold_train_inputs.append(X_train_scaled)
    fold_val_inputs.append(X_val_scaled)
    fold_train_targets.append(y_train_fold)
    fold_val_targets.append(y_val_fold)

print(f"Successfully processed and log-transformed data partitions across all {kf.n_splits} folds.")
```

## 18.Model Fitting & Evaluation

I fit the Lasso Regression model on the log-transformed data spaces. The training and validation metrics are computed on the log scale, providing stable calculations that prevent high-value properties from skewing the final cross-validation summary.


```python
from sklearn.linear_model import LassoCV

fold_train_scores = []
fold_val_scores = []

for i in range(5):
    model = LassoCV(cv=5, random_state=42, max_iter=10000)
    model.fit(fold_train_inputs[i], fold_train_targets[i])
    
    train_r2 = model.score(fold_train_inputs[i], fold_train_targets[i])
    val_r2 = model.score(fold_val_inputs[i], fold_val_targets[i])
    
    fold_train_scores.append(train_r2)
    fold_val_scores.append(val_r2)
    
    print(f"Fold {i+1} -> Log-Encoded Train R²: {train_r2:.4f} | Validation R²: {val_r2:.4f}")


print(f"Mean Training R² Score:   {np.mean(fold_train_scores):.4f}")
print(f"Mean Validation R² Score: {np.mean(fold_val_scores):.4f}")
```

## 19.Target-Encoded Results Analysis after Log transformation

Combining Target Encoding with a natural log transformation (`np.log1p`) yielded a massive performance breakthrough. The model achieved a **Mean Validation $R^2$ Score of 0.8526** and a **Mean Training $R^2$ Score of 0.9039**. 

Stabilizing the price variance completely transformed our predictions:
* Folds 1, 2, 4, and 5 are now highly stable, delivering elite validation scores between **0.8856 and 0.9098**.
* **Fold 3 recovered dramatically, climbing from 0.5430 up to 0.6708**, proving that compressing the outlier luxury home prices successfully prevented the model weights from distorting.


# Executive Summary: Ames Housing Price Prediction Pipeline

## 1. Core Objective
The objective was to engineer a robust, end-to-end machine learning pipeline to predict residential real estate prices while systematically resolving structural data type traps, data leakage, and severe variance distortions caused by pricing outliers.

## 2. Analytical Strategy and Pipeline Architecture
Instead of applying standard, blunt-force imputation or arbitrary categorical encoding across the entire dataset, the engineering process was broken down into five core phases:

* **Rule-Based Data Restoration:** Identified features where "NA" explicitly represented a physical structural absence (e.g., no basement, pool, or alley) and safely renamed them to 'None'. This isolated them from true data-entry omissions (like missing LotFrontage), which were handled strictly with statistical medians.
* **Data Type Correction:** Converted MSSubClass from arbitrary integers into string attributes (object). This stopped the model from treating dwelling codes as continuous mathematical scales.
* **Leakage-Free Feature Processing:** Transitioned away from dataset-wide One-Hot Encoding (which bloated the matrix to 245 sparse features). Instead, Target Encoding was implemented to map categories directly to their historic pricing impact. To enforce perfect data isolation, all categorical averages, imputations, and variance scaling were computed locally inside individual cross-validation folds.
* **Target Variance Stabilization:** Applied a natural logarithmic transformation (np.log1p) to the target variable (SalePrice) to compress the heavily right-skewed tail caused by luxury real estate outliers.
* **Regularized Model Training:** Deployed a Lasso Regression estimator inside a 5-fold cross-validation routine. Lasso introduced an L1 regularization penalty around the features, actively shrinking the weights of non-essential variables down to absolute zero to perform automated feature selection.

## 3. Performance and Optimization Metrics
The evolutionary performance across each iteration highlights the massive impact of these strategic optimization decisions:

| Modeling Strategy | Mean Training R² | Mean Validation R² | Fold 3 Validation Result |
| :--- | :--- | :--- | :--- |

| **Log-Transformed Target Encoded Lasso** | **0.9039** | **0.8526** | **0.6708 (Successfully recovered from variance drops)** |

## 4. Conclusion and Core Insights
By introducing the target log-transformation alongside fold-isolated target encoding, the final model achieved an elite Mean Validation R² score of 0.8526, with individual peak folds consistently hitting 0.88 to 0.90. This unified framework neutralized extreme real estate outliers, stabilized model coefficients, eliminated data leakage, and delivered a highly predictive corporate-grade model that outpaces standard baseline regressions.
